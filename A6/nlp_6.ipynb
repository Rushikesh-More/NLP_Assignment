```python
from transformers import MarianMTModel, MarianTokenizer
import torch
```

```python
# Model name for English to Hindi translation
model_name = "Helsinki-NLP/opus-mt-en-hi"

# Load tokenizer and model
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

print("Model loaded successfully!")
```

```python
def translate_text(text):

    # Tokenize the input sentence
    tokens = tokenizer(text, return_tensors="pt", padding=True)

    # Generate translation
    translated = model.generate(**tokens)

    # Decode the translated tokens
    output = tokenizer.decode(translated[0], skip_special_tokens=True)

    return output
```

```python
english_sentence = "India is a diverse country with many languages."

hindi_translation = translate_text(english_sentence)

print("English:", english_sentence)
print("Hindi:", hindi_translation)
```

```python
sentences = [
    "Welcome to India.",
    "Public information should be available in multiple languages.",
    "Machine translation helps people understand different languages."
]

for s in sentences:
    print("English:", s)
    print("Hindi:", translate_text(s))
    print()
```

```python
text = input("Enter English sentence: ")

translation = translate_text(text)

print("Translated Hindi Text:")
print(translation)
```
